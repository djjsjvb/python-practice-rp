```python
import numpy as np
import pandas as pd
```

```python
dmd=pd.read_csv('diamonds.csv')
```

```python
print(dmd.shape)
```

```python
#1 Load the dataset and display the first 10 rows.
dmd.head(10)
```

```python
#2 Check the shape and column names of the dataset
print(dmd.shape)
dmd.columns
```

```python
#3 Find and fill missing numeric values with their respective column means
dmd.loc[2:10,'depth']=np.nan
# print("after insertion of nan values")
# dmd.head(20)
# print("after insertion of the mean")
dmd['depth']=dmd['depth'].fillna(dmd['depth'].mean())
dmd.head(20)      #DOUBTS
```

```python
#4 Filter rows where `cut` is "Ideal" and `color` is "E".
sec_dmd=dmd[(dmd['cut']=='Ideal')&(dmd['color']=='E')]
display(sec_dmd)
print(sec_dmd.shape)
```

```python
#5 Group by `cut` and compute the average `price` for each group.
thd_dmd=dmd.groupby(['cut'])['price'].mean()
display(thd_dmd)
```

```python
#6 Sort the data by `carat` in descending order and show the top 5 rows.
for_dmd=dmd.sort_values(['carat'],ascending=False)
for_dmd.head(20)
```

```python
#7 Convert the `price` column to a NumPy array and find its mean, median, and standard deviation.
dmd['price'].describe()# describe function is used to find the values of the mean, median, standard deviation
# DOUBT
```
